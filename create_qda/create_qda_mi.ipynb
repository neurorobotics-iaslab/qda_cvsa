```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
from sklearn.model_selection import cross_val_score
from sklearn.metrics import confusion_matrix, balanced_accuracy_score
from sklearn.model_selection import GroupKFold, GridSearchCV, GroupShuffleSplit
from sklearn.model_selection import train_test_split
import seaborn as sns
import os
import math
```

# import data

```python
def fix_mat(data):
    if data.dtype.names:
        new_data = dict()
        for name in data.dtype.names:
            new_data[name]=data[0][name][0]
        for k,v in new_data.items():
            if v.dtype.names:
                new_data[k] = fix_mat(v)
            else:
                new_data[k] = np.squeeze(v)
        return new_data
    else:
        return data
```

```python
data_name = 'data_c7_21012026_171955_mi_dummy' # to do metti le ore del giorno in cui alleni
cluster_name = 'gmm'
path = './datasets/' + cluster_name + '/' + data_name + '.mat'
data = loadmat(path)
bands = fix_mat(data['bands'])
classes = fix_mat(data['classes'][0])
percentual_test = 0.25
idx_channels = fix_mat(data['idx_channels']).ravel() - 1  # -1 for python indexing  ------> index choosen for the qda
channels_labels = fix_mat(data['channels_labels']).ravel()  
files = fix_mat(data['filenames'])
subject = data_name[5:7] # fixed position for the subject code
day = data_name[8:] # fixed position for the day code
model_used_for_dataset = fix_mat(data['gmm_file'])
groups = fix_mat(data['trials']).ravel()

X = fix_mat(data['X'])
X = np.log(X) ######################## log-transform the data
features_list = []
for i in range(X.shape[1]):
    indices = idx_channels[i].ravel().astype(int) 
    band_features = X[:, i, indices]
    features_list.append(band_features)
X = np.hstack(features_list)
y = fix_mat(data['y']).ravel()

gss = GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=42) # divide the data ensuring that trials are not split between train and test
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
groups_train = groups[train_idx]  

print(f"Total samples: {len(X)}, channels: {X.shape[1]}")
print(f"Samples for Train: {len(X_train)} (Trial: {len(np.unique(groups_train))}), 769: {sum(y_train==classes[0])}, 770: {sum(y_train==classes[1])}")
print(f"Samples for Test: {len(X_test)} (Trial: {len(np.unique(groups[test_idx]))}), 769: {sum(y_test==classes[0])}, 770: {sum(y_test==classes[1])}")
```

# Usefull functions

```python
def print_cm(cm, title, tick_labels):
    # Create a heatmap
    plt.figure(figsize=(3,2))
    ax = sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    ax.set_xticklabels(tick_labels)
    ax.set_yticklabels(tick_labels)
    
    # Add labels and titles
    plt.xlabel('Predicted Labels')
    plt.ylabel('True Labels')
    plt.title(title)
    plt.show()
    

def print_cm_rej(model, X, y_true, th_rej, title, ticket_label, type=' balanced'):
    cm = [[0, 0], [0, 0]]
    y_probs = model.predict_proba(X)
    y_pred = model.predict(X)
    count_integrated = 0
    
    for idx_p in range(len(y_probs)):
        if y_true[idx_p] == 769 and y_probs[idx_p,0] > th_rej[0]:
            cm[0][0] += 1 
            count_integrated = count_integrated + 1
        elif y_true[idx_p] == 769 and y_probs[idx_p, 1] > th_rej[0]:
            cm[0][1] += 1
            count_integrated = count_integrated + 1
        elif y_true[idx_p] == 770 and y_probs[idx_p, 1] > th_rej[1]:
            cm[1][1] += 1
            count_integrated = count_integrated + 1
        elif y_true[idx_p] == 770 and y_probs[idx_p, 0] > th_rej[1]:
            cm[1][0] += 1
            count_integrated = count_integrated + 1

    # print accuracy
    print('Accuracy:', np.trace(cm)/np.sum(cm))
    print(f"Balanced accuracy:     {balanced_accuracy_score(y_true, y_pred):.4f}")
    print('Accuracy 769:', cm[0][0]/(cm[0][0]+cm[0][1]+1e-10))
    print('Accuracy 770:', cm[1][1]/(cm[1][0]+cm[1][1]+1e-10))
    print('Integrated: ', count_integrated, ' percentual: ', count_integrated/len(y_probs), '%')   
    print_cm(cm, title, ticket_label)
    
    
def histo_prediction(X, y, model, bins_step, classes):
    probs = model.predict_proba(X)
    y_pred = model.predict(X)

    correct_mask = (y_pred == y)
    incorrect_mask = (y_pred != y)

    probs_class_0 = probs[:, 0]
    
    probs_correct = probs_class_0[correct_mask]
    probs_incorrect = probs_class_0[incorrect_mask]

    bins = np.arange(0, 1.0 + bins_step, bins_step)

    figure = plt.figure(figsize=(8, 5)) 
    plt.hist(
        [probs_correct, probs_incorrect], 
        bins=bins, 
        stacked=True, 
        color=['#28a745', '#dc3545'], # Color: green (correct), red (wrong)
        label=['Correctly classified', 'Wrongly classified']
    )

    plt.xticks(np.arange(0, 1.1, 0.1))
    plt.title(f'Probability distribution class: {classes[0]}')
    plt.xlabel(f'Probability class {classes[0]}')
    plt.ylabel('Frequency')
    plt.legend() 

    plt.show()
    
```

# QDA

```python
ths_rej = [0.5, 0.5]
bins_step = 0.02

param_grid = {
    'reg_param': [0.001, 0.01, 0.05, 0.1, 0.15, 0.2, 0.25, 0.3, 0.4, 0.5], 
    'priors': [[0.5, 0.5]]
}

print("--- Start Grid Search ---")
gkf = GroupKFold(n_splits=5)

grid_search = GridSearchCV(
    estimator=QuadraticDiscriminantAnalysis(), 
    param_grid=param_grid, 
    cv=gkf, 
    n_jobs=-1, 
    scoring='balanced_accuracy'
).fit(X_train, y_train, groups=groups_train)

model_qda = grid_search.best_estimator_
best_params = grid_search.best_params_

y_pred_train = model_qda.predict(X_train)
y_pred_test  = model_qda.predict(X_test)

score_train = balanced_accuracy_score(y_train, y_pred_train)
score_test  = balanced_accuracy_score(y_test, y_pred_test)

print(f"\nBest param for reg_param = {best_params['reg_param']}")
print("-" * 40)
print(f"1. Score CV (Mean 5-fold):     {grid_search.best_score_:.4f}")
print(f"2. Score Train (Balanced):     {score_train:.4f}")
print(f"3. Score Test (Balanced):      {score_test:.4f}")
print("-" * 40)

print("--- Best model GS results ---")
print("------ Train set ------")
print_cm_rej(model_qda, X_train, y_train, ths_rej, 'train set',  ['769', '770'])
histo_prediction(X_train, y_train, model_qda, bins_step, classes)
print("-" * 40)
print("------ Test set ------")
print_cm_rej(model_qda, X_test,  y_test,  ths_rej, 'test set',  ['769', '770'])
histo_prediction(X_test, y_test, model_qda, bins_step, classes)
print("-" * 40)
print("------ All dataset ------")
print_cm_rej(model_qda, X,  y,  ths_rej, 'all dataset',  ['769', '770'])
histo_prediction(X, y, model_qda, bins_step, classes)

```

```python
print("\n--- Training on all data ---")

final_reg_param = best_params['reg_param']

model_qda = QuadraticDiscriminantAnalysis(
    reg_param=final_reg_param, 
    priors=[0.5, 0.5],
    store_covariance=True 
)
model_qda.fit(X, y)

print("------ Final model results ------")
print_cm_rej(model_qda, X_train, y_train, ths_rej, 'train set',  ['769', '770'])
histo_prediction(X_train, y_train, model_qda, bins_step, classes)
```

```python
# save qda model for sklearn
from ruamel.yaml import YAML
from ruamel.yaml.scalarstring import LiteralScalarString

files_str = []
for i in range(0, len(files[0])):
    str_file = files[0][i]
    files_str.append(str(str_file))
    

model_params = {
    'QdaCfg': {
        'name': "qda",
        'params': {
            'subject': subject,
            'filename': 'qda',
            'filename_dataset': path,
            'files_dataset': files_str,
            'model_used_for_dataset': str(model_used_for_dataset[0]),
            'nclasses': int(len(model_qda.classes_)),
            'classes': model_qda.classes_.tolist(),
            'training_accuracy': float(model_qda.score(X, y)),
            'nfeatures': int(model_qda.n_features_in_),
            'note': 'idchannels are saved in matlab convention, instead channels are the used one (=channels_labels(idchanels in matlab))',
            'channels': [[str(x[0]) for x in band.ravel()] for band in channels_labels],
            'idchannels': [(arr.ravel().astype(int) + 1).tolist() for arr in idx_channels], 
            'bands': [b.ravel().tolist() for b in fix_mat(data['bands'][0]).tolist()],
            'reg_param': float(model_qda.reg_param),
            'priors': model_qda.priors_.tolist(),
            'means': model_qda.means_.tolist(),
            'covs': [arr.tolist() for arr in model_qda.covariance_],
            'rotations': [arr.tolist() for arr in model_qda.rotations_],
            'scalings': [arr.tolist() for arr in model_qda.scalings_]
        }
    }
}

# Save parameters to a YAML file
c_dir = os.getcwd()
c_prev = os.path.abspath(os.path.join(c_dir, os.pardir))
path2save = c_prev + '/cfg/qda_' + subject + '_' + day + '.yaml'

yaml = YAML()
with open(path2save, 'w') as file:
    yaml.dump(model_params, file)
    
print(path2save)
```
